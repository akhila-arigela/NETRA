# Network Traffic Volume and Anomalies

```python
from scipy.stats import ttest_ind

normal = df[df['binary_target']==0]
attack = df[df['binary_target']==1]

# Source Bytes
t_stat, p_value = ttest_ind(
    normal['srcbytes'],
    attack['srcbytes'],
    equal_var=False
)

print("Source Bytes")
print("T-statistic:",t_stat)
print("P-value:",p_value)

# Destination Bytes
t_stat, p_value = ttest_ind(
    normal['dstbytes'],
    attack['dstbytes'],
    equal_var=False
)

print("\nDestination Bytes")
print("T-statistic:",t_stat)
print("P-value:",p_value)
```
